# Date, time and datetime module assignment

# date

```python
from datetime import date

today = date.today()

print(today.strftime("%d:%m:%Y"))
print(today.day)
print(today.month)
print(today.year)
```

```python
# time

import time

current = time.strftime("%H:%M:%S")
print(current)

time.sleep(5)
print("done")
```

```python
# time
# 1 Print the current timestamp.
import time
current_time = time.strftime("%H:%M:%S")

# Wait for 5 seconds, then print "Done".
# time.sleep(5)
print(current_time)

# Print the current local time.
local = time.localtime()
print(time.strftime("%d-%m-%Y %H:%M:%S",local))


# Print only the current time in HH:MM:SS.
current_time = time.strftime("%H:%M:%S")
print("Current time: ",current_time)

# Print the readable formate time
read_time = time.ctime()
print(read_time)

# Print the current weekday name.
weekday = time.strftime("%A")
print("weekday: ",weekday)

```

```python
# date time

from datetime import datetime

#Print the current date and time.
ct_date_time = datetime.now()
print(ct_date_time)

# Print only today's date in DD-MM-YYYY.
today = date.today()
print(today.strftime("%D:%M:%Y"))
```

```python
from datetime import datetime
current = datetime.now() # To find the current time

new = current.replace(year=2030) # replaced current year with 2030

print(new.strftime("%D:%M:%Y")) 
```
