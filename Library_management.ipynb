## 3. Search functions

```python
# Sub-Functions for search menu

def search_by_id(library, book_id):
    for book in library:
        if book["book_id"] == book_id:
            return [book]
    
    
# Function for searching by book title:

def search_by_title(library, title):
    search_books = []
    for book in library:
        if title.lower() in book["title"].lower() :
            search_books.append(book)
    
    return search_books

# Function for searching by book title:

def search_by_author(library, author):
    search_books = []
    for book in library:
        if author.lower() in book["author"].lower():
            search_books.append(book)

    return search_books
```

```python
# Search Menu function

library = [
    {'book_id': '100', 'title': 'Harry Potter', 'author': 'Mohit', 'available': True, 'assigned_to': None, 'date_added': '18/07/26 at 04:28 PM'},
    {'book_id': '102', 'title': 'Amazing Spider Man', 'author': 'Krishna', 'available': True, 'assigned_to': None, 'date_added': '18/07/26 at 04:29 PM'},
    {'book_id': '103', 'title': 'Inception', 'author': 'Dr Chotu', 'available': True, 'assigned_to': None, 'date_added': '18/07/26 at 04:29 PM'}
]

# Function for search menu

def search_menu(library):
    while True:

        # Options for search menu
        print("="*10, "Search a book", "="*10)
        print("a. By ID")
        print("b. By Title")
        print("c. By Author")
        print("d. back \n")

        # Taking user input
        choice = input("select the option: ").lower()

        # controll flow for user choice
        if choice=="a":
            print("Enter a book ID")
            search_id = input("Enter book id: ")
            book = search_by_id(library,search_id)

            # output 
            return book

        elif choice=="b":
            search_title = input("Enter book title: ").title()
            books = search_by_title(library,search_title)  

            # output 
            return books

        elif choice=="c":
            search_author = input("Enter book author: ").title()
            books_2 = search_by_author(library,search_author)

            # output 
            return books_2

        elif choice=="d":
            break

        else:
            print("Enter valid input")
```

## 4. Book Management Functions: ADD, EDIT, DELETE, BOOK MANAGEMENT

```python
# Function to add book

from datetime import datetime

def add_book():
    print(f"{"="*10} Add Book {"="*10}")
    
    # input for the user to add books
    book_id = str(input("Book ID: ")).title()
    title = str(input("Title: ")).title()
    author = str(input("Author: ")).title()
    available = True
    assigned_to = None
    date_added = datetime.now().strftime(f"%d/%m/%y at %I:%M %p")

    # book dictionary
    book = {
        "book_id": book_id,
        "title": title,
        "author": author,
        "available": available,
        "assigned_to": None,
        "date_added": date_added
    }
    return book
```

```python
# Function to edit a book

def update_book(book):

    print("="*10, "UPDATE BOOK", "="*10)
    while True:
        # if none book is selected
        if not book: 
            break

        # if book is selected
        else:
            # Options to edit book
            print("a. Book Title")
            print("b. Book Author")
            print("c. Availability")
            print("d. back")

            # user choice
            select_field = input("Enter a choice: ")

            # controll flow for selected field
            if select_field == "a":
                update_title = str(input("Enter new title: ")).title()
                book['title'] = update_title
                print("Title updated successfully")
                return book
                

            elif select_field == "b":
                update_author = str(input("Enter new author name: ")).title()
                book['author'] = update_author
                print("Author name updated successfully")
                return book

                

            elif select_field == "c":
                update_availabilty = str(input("Press 'Y' for yes/'N' for No: "))

                # controll flow for availabilty update
                while True:
                    if update_availabilty == "Y":
                        book['available'] = True
                        print("Book is available")
                        break

                    elif update_availabilty == "N":
                        book['available'] = False
                        print("Book is not available")
                        break

                    else:
                        print("Enter a valid option")
                        continue
                return book

            elif select_field == "d":
                break

            else:
                print("Select a valid option")
                continue
```

```python
# Book Management Menu Function

from datetime import datetime


def book_management_menu(library):

    # loop for book management
    while True:

        # Options for book management
        print("="*10, "Book Management", "="*10)
        print("a. Add a Book")
        print("b. Edit a Book")
        print("c. Delete a Book")
        print("d. back \n")
        

        # Taking user input
        choice = input("select the option: ")

        # Control flow of user choice
        if choice == "a":
            print("Enter Book Details")
            book = add_book()  # Function of adding a book
            library.append(book)
            print("Book Added successfully \n")
            

        elif choice == "b":
            print(f"{"="*10} Edit Book {"="*10}")
            book = search_menu(library) # to search for the book
            selected_book = select_book(book) # to select a book from the searched books
            
            print("Update book details")
            print_book(selected_book)

            updated_book = update_book(selected_book)
            print_book(updated_book)

            
        elif choice == "c":
            print(f"{"="*10} Delete Book {"="*10}")
            book = search_menu(library) # to search for the book
            selected_book = select_book(book) # to select a book from the searched books

            # show deleted book
            print_book(selected_book)

            # confirmation to delete a book
            confirm = str(input("Press Y to delete or press any key to exit")).lower()

            # controll flow
            if confirm == "y":
                library.remove(selected_book)
                print("Book deleted successfully \n")
                

            else:
                continue

            
        elif choice == "d":
            break
        else:
            print("Select a valid option")
```

## 5. Selection Functions

```python
# this function will be used for book selection

def select_book(books):

    if not books:
        return None

    elif len(books) == 1:
        
        return books[0]
    
    else:
        for book in books:
            print_book(book)

        # loop for asking user to select book from the searched books

        while True:
            # asking user to select a book
            selected_id = int(input("Enter Book ID to select book"))


            #to return and select the id from the list
            selected_book = search_by_id(books,selected_id)

            # controll flow for book selection
            if selected_book == None:
                print("Enter Valid ID")
                continue
            else:
                return selected_book
```

```python
library = [
    {'book_id': '100', 'title': 'Harry potter', 'author': 'Author mohit', 'available': True, 'assigned_to': None, 'date_added': '18/07/26 at 04:28 PM'},
    {'book_id': '102', 'title': 'amazing spider man', 'author': 'Author krishna', 'available': True, 'assigned_to': None, 'date_added': '18/07/26 at 04:29 PM'},
    {'book_id': '103', 'title': 'inception', 'author': 'Dr chotu', 'available': True, 'assigned_to': None, 'date_added': '18/07/26 at 04:29 PM'}
]

print("="*10, "LIBRARY", "="*10)

while True:

    print("Enter a choice")
    # Main menu options
    print("1. View library")
    print("2. Book management")
    print("3. Search book")
    print("4. Book Allotment")
    print("5. Exit")
    print()


    # taking user input with error handling
    try:
        choice = int(input("Select the option"))
    except ValueError:
        print("Invalid input! Please enter a number.\n")
        continue  # restarts the loop to ask again

    if choice == 1:
        display_books(library)
        
    elif choice == 2:
        book_management_menu(library) 

    elif choice == 3:
        books = search_menu(library)
        for book in books:
            print_book(book)
    
    elif choice == 4:
        # looping the menu option
        while True:
            # header of book
            print(f"{"="*10} BOOK ALLOOTMENT {"="*10} \n")
            print("a. Assign a Book")
            print("b. Return a book")
            print("c. back")
            
            # Asking user to select a option
            select = str(input("Enter an option"))
            
            # Controll flow for user choice
            if select.lower() == "a":
                search_book = search_menu(library)   # to search a book user want to assign
                selected_book = select_book(search_book) # to select a book from the searched books
                print_book(selected_book)

                # Controll flow for Checking book available
                if selected_book["available"] == True:

                    # taking user input for borrower
                    borrower_name = str(input("Enter borrower's name: ")).title()
                    
                    # Updating the book issued to 
                    selected_book["assigned_to"] = borrower_name
                    selected_book["available"] = False
                    print(f"Book assigned to {borrower_name} \n")
                    print_book(selected_book)
                    break
     
 
                else:
                    print("Book is not available")
                    
    
    
            elif select.lower() == "b":

                search_book = search_menu(library)   # to search a book user want to assign
                selected_book = select_book(search_book) # to select a book from the searched books
                print_book(selected_book)

                # Controll flow for Checking book available
                if selected_book["available"] == True:
                    print("Book is not issued to anyone")


                else:
                    print("This book is issued to: ",selected_book["assigned_to"])
                    confirmation = str(input("Confirm return? (Y/N): ")).title()

                    # controll flow for Confirm return
                    if confirmation == "Y":
                        selected_book["assigned_to"] = None
                        selected_book["available"] = True
                        print("Book Returned Successfully")
                    else:
                        break

            elif select.lower() == "c":
                break

            else:
                print("Enter a valid input")
    
    elif choice == 5:
        break
    
    else:
        print("Select a valid option")
                    
```
